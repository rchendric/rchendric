### Zipping all Python Notebooks, Files, and Datasets

Click Run ➡️ Run All Cells (or just execute the following cell!). You'll find `all_my_python_files.zip` in the `/python/` directory on the left side. There are a lot of files, so this might take a minute to run!

*Note:* this notebook should be saved in the `/python/` directory (not in any subdirectory!)

```python
import os
import zipfile


current_directory = os.getcwd()
subdirectories = [f'SkillBuilder-0{ii}' for ii in range(1,10)] + ['LiveLab',
                                                                 'Dr Who Project',
                                                                 'Grammys - Website Analytics Project',
                                                             'Starbucks - RFM Analysis Project']
with zipfile.ZipFile('all_my_python_files.zip', 'w', zipfile.ZIP_DEFLATED) as zipf:
    for subdir in subdirectories:
        subdir_path = os.path.join(current_directory, subdir)
        for root, dirs, files in os.walk(subdir_path):
            for file in files:
                file_path = os.path.join(root, file)
                zipf.write(file_path, os.path.relpath(file_path, current_directory))

print('done 🤠')
```
