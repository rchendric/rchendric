## Reset
Probably the most important one, will revert back to original for most cases
Make sure if testing code, run all cells!

```python
reset = '\u001b[0m'
```

## Colors

```python
# Red
red = '\u001b[31m'

#Black
black = '\u001b[30m'

#Green
green = '\u001b[32m'

#Yellow
yellow = '\u001b[33m'

#Blue
blue = '\u001b[34m'

#Magenta
magenta = '\u001b[35m'

#Cyan
cyan = '\u001b[36m'

#White
white = '\u001b[37m'

##BRIGHT COLORS
#These colors are more bold/stand out even more
#Bright Red
bred = '\u001b[31;1m'

#Bright Black
bblack = '\u001b[30;1m'

#Bright Green
bgreen = '\u001b[32;1m'

#Bright Yellow
byellow = '\u001b[33;1m'

#Bright Blue
bblue = '\u001b[34;1m'

#Bright Magenta
bmagenta = '\u001b[35;1m'

#Bright Cyan
bcyan = '\u001b[36;1m'

#Bright White
bwhite = '\u001b[37;1m'
```

## Color Example

```python
print('Basic colors')
print(f'{black}Black  \n{red}Red  \n{green}Green  \n{yellow}Yellow  \n{blue}Blue  \n{magenta}Magenta  \n{cyan}Cyan  \n{white}White')
print(f'{reset}Bright Colors')
print(f'{bblack}Bright Black  \n{bred}Bright Red  \n{bgreen}Bright Green  \n{byellow}Bright Yellow  \n{bblue}Bright Blue  \n{bmagenta}Bright Magenta  \n{bcyan}Bright Cyan  \n{bwhite}Bright White')
```

## Backgrounds 
Basically a highlight tool!

```python
#Black Background
backblack = '\u001b[40m'

#Red Background
backred = '\u001b[41m'

#Green Background 
backgreen = '\u001b[42m'

#Yellow Background
backyellow = '\u001b[43m'

#Blue Background
backblue = '\u001b[44m'

#Magenta Background
backmagenta = '\u001b[45m'

#Cyan Background 
backcyan = '\u001b[46m'

#White Background
backwhite = '\u001b[47m'


#Bright Backgrounds
#Black Background
bbackblack = '\u001b[40;1m'

#Red Background
bbackred = '\u001b[41;1m'

#Green Background 
bbackgreen = '\u001b[42;1m'

#Yellow Background
bbackyellow = '\u001b[43;1m'

#Blue Background
bbackblue = '\u001b[44;1m'

#Magenta Background
bbackmagenta = '\u001b[45;1m'

#Cyan Background 
bbackcyan = '\u001b[46;1m'

#White Background
bbackwhite = '\u001b[47;1m'
```

## Background Examples
Note:Bright backgrounds make the words printed more noticable, not a different color

```python
print('Basic Backgrounds')
print(f'{backblack} a   \n{backred} a   \n{backgreen} a   \n{backyellow} a   \n{backblue} a   \n{backmagenta} a   \n{backcyan} a   \n{backwhite} a   ')
print(f'{reset}Bright Backgrounds')
print(f'{bbackblack} a   \n{bbackred} a   \n{bbackgreen} a   \n{bbackyellow} a   \n{bbackblue} a   \n{bbackmagenta} a   \n{bbackcyan} a   \n{bbackwhite} a   ')
```

## Decorations

```python
#Bold
bold = '\u001b[1m'

#Underline
underline = '\u001b[4m'

#Reversed
reversed = '\u001b[7m'
```

## Decoration examples

```python
print(f'{bold}Bold \n{reset}{underline}Underline\n{reset}{reversed}Reversed')
print(f'{reset}Can be combined together')
print(f'{bold}{underline}{reversed}woah')
```
